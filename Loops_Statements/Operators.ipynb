```python
index_count = 0
```

```python
word = 'abcde'
```

```python
for letter in word:
    print(f"Letter at index {index_count} is {letter}")
    index_count += 1
```

```python
for item in enumerate(word):
    print(item)
```

```python
for index, letter in enumerate(word):
    print(index)
    print(letter)
```
