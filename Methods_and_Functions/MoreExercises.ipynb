```python
#Write a Python function that accepts a string and calculates the number of upper case letters and lower case letters.
```

```python
def ex3(string):
    lower_case_count = 0
    upper_case_count = 0
    for char in string:
        if char.isupper():
            upper_case_count += 1
        elif char.islower():
            lower_case_count += 1
    return print(f"No. of Upper case characters: {upper_case_count}\nNo. of Lower case characters: {lower_case_count}")
```

```python
ex3('Hello Mr. Rogers, how are you this fine Tuesday?')
```

```python
#Write a Python function to multiply all the numbers in a list.
```

```python
def ex5(numbers):
    sum_nums = 1
    for number in numbers:
        sum_nums = sum_nums * number
    return sum_nums
```

```python
ex5([1, 2, 3, -4])
```

```python
#Write a Python function that checks whether a word or phrase is palindrome or not.
```

```python
def ex6(word):
    no_empty_word = word.replace(' ','')
    reversed_word = no_empty_word[::-1]
    return reversed_word == no_empty_word
```

```python
ex6('helleh')
```

```python
#Write a Python function to check whether a string is pangram or not. (Assume the string passed in does not have any punctuation)
#example : "The quick brown fox jumps over the lazy dog"
```

```python
def ex7(string):
    words_list = string.split()
    chars_list = []
    for word in words_list:
        for letter in word:
            chars_list.append(letter.lower())
    chars_list.sort()
    unique_chars = set(chars_list)
    return len(unique_chars) == 26
```

```python
ex7("The quick brown fox jumps over the lazy dog")
```
