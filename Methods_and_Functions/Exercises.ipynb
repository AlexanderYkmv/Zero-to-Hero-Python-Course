```python
#Define a function called myfunc that takes in a string, and returns a matching string where every even letter is uppercase, 
#and every odd letter is lowercase.Assume that the incoming string only contains letters, and don't worry about numbers, spaces or punctuation. 
#The output string can start with either an uppercase or lowercase letter, so long as letters alternate throughout the string.
```

```python
def myfunc(string):
    new_string = ''
    for index, letter in enumerate(string):
        if index % 2 == 0:
            new_string += letter.upper()
        else:
            new_string += letter.lower()
    return new_string
```

```python
myfunc('Abcdefgh')
```

```python
#Write a function that returns the lesser of two given numbers if both numbers are even, but returns the greater if one or both numbers are odd
```

```python
def ex1(num1,num2):
    if num1 % 2 == 0 and num2 % 2 == 0:
        return min(num1,num2)
    else:
        return max(num1,num2)
```

```python
#Write a function takes a two-word string and returns True if both words begin with same letter
```

```python
def ex3(string):
    first_letters = []
    check_list = string.split(' ')
    for word in check_list:
        first_letters.append(word[0].lower())
    if first_letters[0] == first_letters[1]:
        return True
    else:
        return False
```

```python
ex3('Levelheaded Llama')
```

```python
#Write a function that capitalizes the first and fourth letters of a name
```

```python
def ex4(name):
    new_string = ''
    for index,letter in enumerate(name):
        if index == 0 or index == 3:
            new_string += letter.upper()
        else:
            new_string += letter.lower()
    return new_string
```

```python
ex4("macdonald")
```

```python
#Given a sentence, return a sentence with the words reversed,  'I am home' --> 'home am I'
```

```python
def ex5(sentence):
    words_list = sentence.split(' ')
    reversed_sentence = ''
    reversed_list = reversed(words_list)
    reversed_sentence = " ".join(reversed_list)
    return reversed_sentence
```

```python
#Given three integers 1-11, if their sum is less than or equal to 21, return their sum. 
#If sum > 21 and there's an eleven, reduce the total sum by 10. Finally, if the sum (even after adjustment) exceeds 21, return 'BUST'
#ex9(5,6,7) --> 18    ex9(9,9,9) --> 'BUST'      ex9(9,9,11) --> 19
```

```python
def ex9(num1,num2,num3):
    sum_nums = num1 + num2 + num3
    if sum_nums <= 21:
        return sum_nums
    elif sum_nums > 21 and num1 == 11 or num2 == 11 or num3 == 11:
        sum_nums -= 10
    if sum_nums > 21:
        return 'BUST'
    return sum_nums
```

```python
ex9(5,6,7)
```

```python
ex9(9,9,9)
```

```python
ex9(9,9,11)
```

```python
def ex11(list_nums):
    is_spy_list = []
    spy_str = ''
    for number in list_nums:
        if number == 0 or number == 7:
            is_spy_list.append(str(number)) 
    spy_str = "".join(is_spy_list)
    if '007' in spy_str:
        return True
    else:
        return False
```
