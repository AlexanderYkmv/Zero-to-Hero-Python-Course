```python
def square(number):
    return number ** 2
```

```python
my_list = [1,2,3,4,5]
```

```python
map(square, my_list)   #map syntax:(function, iterable object)
```

```python
for item in map(square,my_list):
    print(item)
```

```python
list(map(square,my_list))
```

```python
def check_even(num):
    return num % 2 == 0
```

```python
new_list = [1,2,3,4,5,6]
```

```python
filter(check_even,new_list)  #filter syntax:(boolean function, iterable object)
```

```python
for item in filter(check_even,new_list):
    print(item)
```

```python
list(map(lambda num:num**2, my_list))
```
