```python
def check_even_list(num_list):
    
    for number in num_list:
        if number % 2 == 0:
            return True
    return False
            
```

```python
check_even_list([1,1,5,3])
```

```python
check_even_list([1,2,5,5])
```

```python
check_even_list([2,4])
```

```python
check_even_list([1,2])
```

```python
empty_list = []
def give_even_numbers(num_list):
    for number in num_list:
        if number % 2 == 0:
            empty_list.append(number)
    return empty_list
```

```python
give_even_numbers([1,2,3,4])
```

```python
def overvalued_stock(stocks_prices):
    max_price = 0
    stock_name = ''
    for ticker,price in stocks_prices:
        if price > max_price:
            max_price = price
            stock_name = ticker
        else:
            pass
    return (stock_name,max_price)
            
```

```python
overvalued_stock(stocks_prices)
```

```python
def my_func(*args):
    even_list = []
    for arg in args:
        if arg % 2 == 0:
            even_list.append(arg)
    return even_list
```

```python
my_func(1,2,3,4,5)
```
