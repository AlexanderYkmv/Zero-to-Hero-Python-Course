```python
from random import shuffle
values = {'Joker':0,'Two':2, 'Three':3, 'Four':4, 'Five':5, 'Six':6, 'Seven':7, 'Eight':8, 
            'Nine':9, 'Ten':10, 'Jack':10, 'Queen':10, 'King':10, 'Ace':11}
suits = ('Hearts', 'Diamonds', 'Spades', 'Clubs')
ranks = ('Two', 'Three', 'Four', 'Five', 'Six', 'Seven', 'Eight', 'Nine', 'Ten', 'Jack', 'Queen', 'King', 'Ace')
joker = 'Joker'
```

```python
class Card:

    def __init__(self,rank,suit=None):
        self.rank = rank
        self.suit = suit
        self.value = values[rank]
        
    def __str__(self):
        if self.rank == 'Joker':
            return f'{self.rank} card'
        else:
            return f'{self.rank} of {self.suit}'
```

```python
class Deck:

    def __init__(self):
        self.all_cards = []
        for rank in ranks:
            for suit in suits:
                for i in range(0,6):
                    created_card = Card(rank,suit)
                    self.all_cards.append(created_card)
        
    def shuffle(self):
        self.joker_card = Card(joker)
        shuffle(self.all_cards)
        self.all_cards.insert(209,self.joker_card)
```

```python
class Player:
    
    def __init__(self,name,balance=None):
        self.balance = balance
        self.name = name


    def bet(self, bet):
        self.balance -= bet
        return self.balance
```

```python
class Dealer:

    def __init__(self):
        self.dealer_hand = []
        self.player_hand = []
        self.cleared_cards = []
    def hit_dealer(self, deck):
        return self.dealer_hand.append(deck.all_cards.pop(0)) 

    def hit_player(self, deck):
        return self.player_hand.append(deck.all_cards.pop(0))

    def sum_player_hand(self):
        card_points = 0
        for card in self.player_hand:
            card_points += card.value
        return card_points

    def sum_dealer_hand(self):
        card_points = 0
        for card in self.dealer_hand:
            card_points += card.value
        return card_points

    def clear_table(self):
        self.cleared_cards.extend(self.dealer_hand)
        self.cleared_cards.extend(self.player_hand)
        self.player_hand.clear()
        self.dealer_hand.clear()
    def ask_for_bet(self):
        self.bet = input("Dealer asks for player bet or 'Leave' to quit the game: ")
        return self.bet

    def deal_cards(self, deck):
        player_card = deck.all_cards.pop(0)
        self.player_hand.append(player_card)
        dealer_card = deck.all_cards.pop(0)
        self.dealer_hand.append(dealer_card)
        
```

```python
def startup_game():
    print("Welcome to Black Jack!")
    player_name = input("How should the dealer call you?")
    print(f"{player_name} how much money do you wish to deposit?")
    balance = input("Please input a number: ")
    while balance.isdigit() == False:
        print("This is not a number!")
        balance = input("Please input a number: ")
    balance = int(balance)
    new_player = Player(player_name,balance)
    print(f"Congrats {player_name}, you successfully deposited {balance}$")
    print("Shuffling deck...")
    game_deck = Deck()
    game_deck.shuffle()
    dealer = Dealer()
    player_bet = dealer.ask_for_bet()
    while player_bet != 'Leave':
        player_bet = int(player_bet)
        new_player.bet(player_bet)
        print(f"Your bet is {player_bet}$")
        print("Dealing cards..")
        for i in range(0,2):
            dealer.deal_cards(game_deck)
        for card in dealer.player_hand:
            print(f"Your card is {card}")
        print(f"You have {dealer.sum_player_hand()}")
        print(f"Dealer Card is {dealer.dealer_hand[0]}")
        hit_or_stand = input("Hit or Stand?")
        while hit_or_stand == 'Hit':
            dealer.hit_player(game_deck)
            for card in dealer.player_hand:
                print(f"Your card is {card}")
            print(f"You have {dealer.sum_player_hand()}")
            print(f"Dealer Card is {dealer.dealer_hand[0]}")
            if dealer.sum_player_hand() > 21:
                print(f"Bust! You got {dealer.sum_player_hand()} and it's over 21")
                break
            hit_or_stand = input("Hit or Stand?")
        if dealer.sum_player_hand() <= 21:
            print(f"You chose to stand with {dealer.sum_player_hand()} points")
            for card in dealer.dealer_hand:
                print(f"Dealer card is {card}")
            while dealer.sum_dealer_hand() <= 16:
                print("Dealer draws...")
                dealer.hit_dealer(game_deck)
                for card in dealer.dealer_hand:
                    print(f"Dealer card is {card}")
            if dealer.sum_dealer_hand() > 21:
                    print(f"Dealer busted! {new_player.name} you win!")
                    new_player.balance += player_bet * 2
            elif dealer.sum_dealer_hand() < dealer.sum_player_hand():
                print("Player wins!")
                new_player.balance += player_bet * 2
            elif dealer.sum_dealer_hand() > dealer.sum_player_hand():
                print("Dealer wins!")
            else:
                print("Draw")
                new_player.balance += player_bet
        dealer.clear_table()
        for card in dealer.cleared_cards:
            print(card)
        print(f"Your current balance is: {new_player.balance}$")
        player_bet = dealer.ask_for_bet()
```
