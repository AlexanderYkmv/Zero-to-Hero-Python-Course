```python
class Line:
    
    def __init__(self,coor1,coor2):
        (self.x1,self.y1)= coor1
        (self.x2,self.y2) = coor2

    def distance(self):
        return ((self.x2 - self.x1)**2 + (self.y2 - self.y1)**2) ** 0.5

    def slope(self):
        return (self.y2-self.y1) / (self.x2-self.x1)
```

```python
coordinate1 = (3,2)
coordinate2 = (8,10)
li = Line(coordinate1,coordinate2)
li.slope()
```

```python
#Problem 2: Create class Cylinder with methods volume and surface area
```

```python
class Cylinder:

    def __init__(self,height,radius):
        self.height = height
        self.radius = radius

    def volume(self):
        return self.height * (self.radius**2) * 3.14

    def surface_area(self):
        return (2 * 3.14 * self.radius) * (self.height + self.radius)
```

```python
c = Cylinder(2,3)
```

```python
c.volume()
```

```python
c.surface_area()
```
