```python
class Animal():

    def __init__(self):
        print("Animal created!")

    def what_am_i(self):
        print(type(self))

    def eat(self):
        print("I am eating")
```

```python
class Dog(Animal):

    def __init__(self):
        print("Dog created!")

    def what_am_i(self):
        print("I am a dog")

    def speak(self):
        print("Woof")
```

```python
new_animal = Animal()
```

```python
new_animal.what_am_i()
```

```python
new_dog = Dog()
```

```python
new_dog.eat()
```

```python
new_dog.what_am_i()
```

```python
new_dog.speak()
```

```python
class Cat(Animal):

    def __init__(self):
        print("Cat created!")

    def what_am_i(self):
        print("I am a cat")

    def speak(self):
        print("Meow")
```

```python
poly_dog = Dog()
```

```python
poly_cat = Cat()
```

```python
poly_dog.speak()
```

```python
poly_cat.speak()
```
