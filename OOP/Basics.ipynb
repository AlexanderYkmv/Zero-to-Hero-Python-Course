```python
class Dog:
    species = 'mammal'
    def __init__(self,breed,age):
        self.breed = breed
        self.age = age

    def bark(self):
        print(f"Woof my breed is {self.breed}")
    
    def bark_number(self, number):
        print(f"Woof my breed is {self.breed} and the number is {number}")
```

```python
my_dog = Dog(breed = 'Golden', age = 2)
```

```python
type(my_dog)
```

```python
my_dog.species
```

```python
my_dog.age
```

```python
my_dog.bark_number(1)
```

```python
my_dog.bark()
```

```python
class Circle():

    pi = 3.14
    def __init__(self,radius):
        self.radius = radius

    def get_circumfrence(self):
        return self.pi * 2 * self.radius
```

```python
my_circle = Circle(2)
```

```python
my_circle.get_circumfrence()
```
