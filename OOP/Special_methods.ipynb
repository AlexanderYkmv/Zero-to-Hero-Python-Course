```python
class Book():

    def __init__(self,author,title,pages):
        self.author = author
        self.title = title
        self.pages = pages

    def __str__(self):
        return f"{self.title} by {self.author} has {self.pages} pages"

    def __len__(self):
        return self.pages
```

```python
book = Book('Alex','Course',200)
```

```python
print(book)
```

```python
len(book)
```
