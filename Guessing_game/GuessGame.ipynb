```python
from random import randint
num_to_guess = randint(1,100)
num_to_guess
guesses_count = 1
previous_guess = None
player_guess = None
player_guess = int(input('Guess the random number from 1-100:'))
if player_guess < 1 or player_guess > 100:
        print("OUT OF BOUNDS!")
elif abs(player_guess - num_to_guess) <= 10:
        print("WARM!")
else:
        print("COLD!")
while player_guess != num_to_guess:
    guesses_count += 1
    previous_guess = player_guess
    player_guess = int(input('Try again!:'))
    if player_guess < 1 or player_guess > 100:
        print("OUT OF BOUNDS!")
    elif abs(player_guess - num_to_guess) < abs(previous_guess - num_to_guess):
        print("WARMER!")
    else:
        print("COLDER!")
print(f"Congrats! It took you only {guesses_count} guesses")
```
