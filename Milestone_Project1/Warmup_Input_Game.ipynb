```python
#User input program should display a lis, have a user choose an index position and an input value
#replace value at chosen index, display list and ask if user wants to play again, play until no is given
#list will always have only 3 items
```

```python
chosen_index = 0
chosen_value = None
game_list = [2,'base',20]
```

```python
def display_list(input_list):
    print(input_list)
```

```python
def startup_greet():
    print("Welcome!")
    print("Please chose which value you want to change and what you want it to be!")
```

```python
def ask_to_play():
    answer = input("Do you wish to play again(Y/N): ")
    return answer
```

```python
def game():
    print(f"This is the current list: {game_list}")
    chosen_index = int(input("Please chose the index: "))
    while chosen_index not in range(0,3):
        print("Inex out of bounds!")
        chosen_index = int(input("Please chose the index: "))
    chosen_value = input("Please chose the value you want to set: ")
    if chosen_value.isdigit():
        chosen_value = int(chosen_value)
    game_list[chosen_index] = chosen_value
    display_list(game_list)
```

```python
def game_launcher():
    startup_greet()
    game()
    want_to_play = ask_to_play()
    while want_to_play == 'Y':
        game()
        want_to_play = ask_to_play()
    print("Thanks for playing!")
```

```python
game_launcher()
```
