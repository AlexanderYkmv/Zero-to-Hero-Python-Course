```python
def hello(name = 'Sasho'):
    print("Hello func executed")

    def greet():
        return "\t Greet inside hello"

    def welcome():
        return "\t Welcome inside hello"
    print("Im going to return a func")
    if name == 'Sasho':
        return greet
    else:
        return welcome
```

```python
hello()
```

```python
new_func = hello("Name")
```

```python
new_func()
```

```python
def say_hi():
    return "Hi"
```

```python
def other(some_func):
    print("Other func")
    print(some_func())
```

```python
other(say_hi)
```

```python
def new_decorator(some_func):

    def func_wrapper():
        print("Additional functionality")
        some_func()

    return func_wrapper
```

```python
def decorate_me():
    print("Boring function")
```

```python
fancy_func = new_decorator(decorate_me)
```

```python
fancy_func()
```

```python
@new_decorator
def decorate_me():
    print("Boring function")
```

```python
decorate_me()
```
