```python
def create_cubes(n):
    result = []
    for x in range(n):
        result.append(x**3)
    return result
```

```python
create_cubes(5)
```

```python
def generator_cubes(n):
    for x in range(n):
        yield x**3
```

```python
for number in generator_cubes(5):
    print(number)
```

```python
def fib_generator(n):
    a = 1
    b = 1
    for number in range(n):
        yield a
        a = b
        b = a + b
```

```python
for number in fib_generator(10):
    print(number)
```

```python
def simple_gen():
    for num in range(3):
        yield num
```

```python
for number in simple_gen():
    print(number)
```

```python
g = simple_gen()
```
