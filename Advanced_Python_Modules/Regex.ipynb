```python
text = "User's phone number is 421-333-5506. Call phone soon!"
```

```python
import re
```

```python
match = re.search('phone',text)
```

```python
match
```

```python
match.span()
```

```python
matches = re.findall('phone', text)
```

```python
matches
```

```python
phone = re.search(r'\d{3}-\d{3}-\d{4}', text)
```

```python
phone
```

```python
phone.group()
```

```python
phone_pattern = re.compile(r'(\d{3})-(\d{3})-(\d{4})')
```

```python
res = re.search(phone_pattern,text)
```

```python
res
```

```python
res.group()
```

```python
res.group(1)
```
